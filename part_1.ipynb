```python
import os
import pandas as pd


# Create the train_15 and train_1500 dataset
orig_data = pd.read_csv("train.csv", index_col=0,nrows=1500)
subset_path = "train_1500.csv"
if not os.path.exists(subset_path):
    orig_data.iloc[:1500,:].to_csv(subset_path)
    print(f"Created {subset_path} with first 1500 rows.")
else:
    print(f"{subset_path} already exists.")


subset_path = "train_15.csv"

if not os.path.exists(subset_path):
    orig_data.iloc[:15].to_csv(subset_path)
    print(f"Created {subset_path} with first 15 rows.")
else:
    print(f"{subset_path} already exists.")

# Create LCSS_Images folder if it doesn't exist
folder_path = "LCSS_Images"
os.makedirs(folder_path, exist_ok=True)
print(f"Folder '{folder_path}' is ready.")
```
